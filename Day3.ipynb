Bivatiate Analysis

Relationship between TWO variables

```python
# groupby(): method in pandas splits a DataFrame into distinct groups based on one or more column values
# Departments vs Average CGPA
df.groupby("Department")["CGPA"].mean()
```

```python
# Department VS Attendance
df.groupby("Department")["Attendance"].mean()
```

```python
#Placement Vs CGPA
df.groupby("Placement_Status")["CGPA"].mean()
```

```python
# Average Study Hours
df.groupby("Department")["Study_Hours"].mean()
```

Multivariate Analysis

Relationship among multiple variables

```python
df.groupby("Department")[["CGPA","Attendance"]].mean()
```

```python
# Crosstab: It provides a tabular summary that shows the frequency
#distribution or raw count of how often combinations of
# specific categories appear together in the dataset.
pd.crosstab(df["Department"],
            df["Placement_Status"])
```

```python
# It determines how strongly and in what direction two variables tend to change together.
# 1.0 (Perfect Positive Correlation): As one variable increases, the other increases proportionally.
# -1.0 (Perfect Negative Correlation): As one variable increases, the other decreases proportionally.
# 0.0 (No Linear Correlation): The movement of one variable does not predict the movement of the other.
corr = df.corr(numeric_only=True)

corr
```
