```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import accuracy_score, mean_squared_error, r2_score
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
df = pd.read_csv("/content/drive/MyDrive/Teen_Mental_Health_Dataset.csv")
```

```python
features = [
    "daily_social_media_hours",
    "sleep_hours",
    "academic_performance",
    "stress_level",
    "anxiety_level",
    "addiction_level"
]

X = df[features]
y = df["depression_label"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

Linear Regression is a supervised learning algorithm used to predict a continuous numerical value by finding the best-fit straight line between the input features and the target variable.

Output Examples

House Price

Salary

Temperature

Sales

Marks

Characteristics

Uses all features.

No penalty on coefficients.

Can overfit if many irrelevant features exist.

Easy to understand.

```python
linear = LinearRegression()
linear.fit(X_train, y_train)
linear_pred = linear.predict(X_test)

print(linear_pred[:10])
linear_class = []

for value in linear_pred:
    if value >= 0.5:
        linear_class.append(1)
    else:
        linear_class.append(0)
print("Linear Regression Accuracy")

print(
    accuracy_score(
        y_test,
        linear_class
    )
)
print("R2 Score")

print(r2_score(y_test, linear_pred))

print("Mean Squared Error")

print(mean_squared_error(y_test, linear_pred))
```

```python
new_data = [[7,5,5.5,8,7,6]]

result = linear.predict(new_data)

print(result)
```

Suppose your dataset has 100 features, but only 20 are important.

Linear Regression may assign very large weights (coefficients) to some features, making the model too sensitive to the training data.

Ridge Regression solves this by penalizing large coefficients

Ridge Regression is a regularized version of Linear Regression that adds an L2 penalty to the cost function. This penalty shrinks coefficient values but does not remove any features.

Characteristics

Reduces overfitting.

Keeps every feature.

Reduces coefficient magnitudes.

Handles multicollinearity better.

| Feature    | Coefficient |
| ---------- | ----------: |
| CGPA       |          25 |
| Attendance |          15 |
| Projects   |          20 |
| Sports     |          12 |

After Applying Ridge

| Feature    | Coefficient |
| ---------- | ----------: |
| CGPA       |          18 |
| Attendance |          11 |
| Projects   |          15 |
| Sports     |           8 |

```python
ridge = Ridge(alpha=1.0)
ridge.fit(X_train, y_train)
ridge_pred = ridge.predict(X_test)
ridge_class = []

for value in ridge_pred:
    if value >= 0.5:
        ridge_class.append(1)
    else:
        ridge_class.append(0)

print("Ridge Accuracy")

print(
accuracy_score(
    y_test,
    ridge_class
))

print("R2 Score")

print(
r2_score(
    y_test,
    ridge_pred
))

print("MSE")

print(
mean_squared_error(
    y_test,
    ridge_pred
))
```

```python
prediction = ridge.predict(new_data)

print(prediction)
```

Sometimes many features are completely unnecessary.

Instead of just reducing their influence, why not remove them?

That is exactly what Lasso does.

Lasso Regression is a regularized version of Linear Regression that adds an L1 penalty to the cost function. It shrinks coefficients and can reduce some of them exactly to zero, automatically performing feature selection.

Characteristics

Reduces overfitting.

Performs feature selection.

Produces simpler, more interpretable models.

Very useful when the dataset contains many irrelevant features.

Before Lasso
| Feature            | Coefficient |
| ------------------ | ----------: |
| CGPA               |          25 |
| Attendance         |          18 |
| Projects           |          20 |
| Sports             |           7 |
| Favorite Color     |           5 |
| Number of Siblings |           4 |

After Lasso
| Feature            | Coefficient |
| ------------------ | ----------: |
| CGPA               |          22 |
| Attendance         |          15 |
| Projects           |          18 |
| Sports             |           0 |
| Favorite Color     |           0 |
| Number of Siblings |           0 |


```python
lasso = Lasso(alpha=0.01)
lasso.fit(
    X_train,
    y_train
)
lasso_pred = lasso.predict(
    X_test
)
lasso_class = []

for value in lasso_pred:
    if value >= 0.5:
        lasso_class.append(1)
    else:
        lasso_class.append(0)
print("Lasso Accuracy")

print(
accuracy_score(
    y_test,
    lasso_class
))

print("R2 Score")

print(
r2_score(
    y_test,
    lasso_pred
))

print("MSE")

print(
mean_squared_error(
    y_test,
    lasso_pred
))
```

```python
prediction = lasso.predict(new_data)

print(prediction)
```

```python
results = pd.DataFrame({

    "Model":[
        "Linear Regression",
        "Ridge Regression",
        "Lasso Regression"
    ],

    "Accuracy":[

        accuracy_score(y_test, linear_class),

        accuracy_score(y_test, ridge_class),

        accuracy_score(y_test, lasso_class)

    ],

    "R2 Score":[

        r2_score(y_test, linear_pred),

        r2_score(y_test, ridge_pred),

        r2_score(y_test, lasso_pred)

    ],

    "MSE":[

        mean_squared_error(y_test, linear_pred),

        mean_squared_error(y_test, ridge_pred),

        mean_squared_error(y_test, lasso_pred)

    ]

})

print(results)
```

Linear Regression: "Uses all features as they are."

Ridge Regression: "Keeps all features but reduces their influence."

Lasso Regression: "Keeps only the important features by removing unnecessary ones."

| Feature                   | Linear Regression         | Ridge Regression         | Lasso Regression                                                 |
| ------------------------- | ------------------------- | ------------------------ | ---------------------------------------------------------------- |
| Algorithm Type            | Regression                | Regression               | Regression                                                       |
| Regularization            | None                      | L2                       | L1                                                               |
| Main Purpose              | Predict continuous values | Reduce overfitting       | Reduce overfitting + Feature selection                           |
| Coefficients              | Original values           | Shrunk toward zero       | Shrunk, some become exactly zero                                 |
| Removes Features          | No                        | No                       | Yes                                                              |
| Handles Multicollinearity | Poorly                    | Well                     | Better than Linear, but can choose one among correlated features |
| Model Complexity          | Higher                    | Medium                   | Lower                                                            |
| Best For                  | Small/simple datasets     | Most features are useful | Many irrelevant features                                         |
