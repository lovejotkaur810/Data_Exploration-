What is Data?
Data is a collection of facts, observations, measurements, or records that can be analyzed to gain insights.

Why Data Types Matter?

Before analyzing data, Python needs to know what kind of information it is handling.

Examples:

Age → Integer                  
Salary → Float                                   
Name → String                                               
Is_Placed → Boolean

These data types determine what operations can be performed.

```python
# Integer (int) - Used for whole numbers.
age = 21
students = 50

print(age)
print(type(age))
```

```python
# Float - Use for decimal values
salary = 55000.75

print(salary)
print(type(salary))
```

```python
# String (str) - stores text
name = "123"

print(name)
print(type(name))
```

```python
# Boolean (bool)- use for true or false
is_placed = True

print(is_placed)
print(type(is_placed))
```

Python Data Structures                                        
Data Analytics rarely works with single values. We work with collections of data

List-




*   Ordered, mutable collections of items.
*   Created using square brackets []
*   Allows duplicate elements and mixed data types.

```python
marks = [78, 85, 90, 67, 88]

print(marks)
print(type(marks))
```

```python
sales = [1200, 1500, 1800, 2200]

print(max(sales))
print(min(sales))
print(sum(sales))
```

Tuples


*   Ordered, immutable sequences of items.
*   Created using parentheses ()
*   Cannot be changed after creation, making them memory-efficient.

```python
months = ("Jan", "Feb", "Mar")

print(months)
print(type(months))
```

Dictionaries (dict)


*   Unordered, mutable key-value maps.
*   Created using curly braces {key: value}
*   Keys must be unique and immutable

```python
user = {"name": "Alex", "role": "Developer"}

print(user)
print(type(user))
```

```python
student = {
    "Name": "Aman",
    "Age": 21,
    "CGPA": 8.4
}

print(student)
print(student["CGPA"])
```

Sets

*   Unordered collections of unique elements
*   Created using curly braces {} or set()
*   Automatically eliminates duplicates; ideal for membership testing.

```python
unique_ids = {101, 102, 103, 101}
print(unique_ids)
print(type(unique_ids))
```

```python
cities = {"Delhi", "Mumbai", "Delhi", "Pune"}

print(cities)
```

Practice Question 3                                                    Write a Python function that takes a list and returns a new list with unique elements of the first list.

```python
def unique_list(l):
  x = []
  for a in l:
    if a not in x:
      x.append(a)
  return x
list1=[1,2,3,3,3,3,4,5]
print(unique_list(list1))
```

Practice Question 10: Write a Python function that takes two lists and returns True if they have at least one common member

Practice Question 10: Write a Python function that takes two lists and returns True if they have at least one common member

```python
list1=['red','green','white','black','pink','yellow']
list2=['a','b','c','d','red']

print("List 1: ")
print(list1)
print("\nList 2: ")
print(list2)
def fun(l1,l2):
    result=False
    for i in l1:
        for j in l2:
            if i==j:
                result=True
    return result

print("\nOutput")
print(fun(list1,list2))
```

Practice Question 11 Program to find no. Of repeated items in the tuple

```python
t1=('a','b','c','a','b','d')
print("\n Tuple is: ")
print(t1)
words=''.join(t1)
l=dict()
for i in words:
	if i not in l:
		l[i]=1
	else:
		l[i]=l[i]+1
print("\n Repeated Items in the tuple are: ")
print(l)
```

Data Transformation

```python
# Add new column "Performance Grade"

def grade(cgpa):
    if cgpa >= 9:
        return "Excellent"
    elif cgpa >=8:
        return "Good"
    elif cgpa >=7:
        return "Average"
    else:
        return "Poor"

df["Grade"] = df["CGPA"].apply(grade)
```

```python
df.head()
```

```python
# Creating attendance status
df["Attendance_Status"] = df["Attendance"].apply(
    lambda x: "Regular" if x>=75 else "Irregular"
)
```
