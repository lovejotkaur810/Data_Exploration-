Methods to Handle the Missing Value


1.   Remove Row
2.   Fill the Mean

1.   Fill the Median
2.   Fill the Mode





```python
# To remove rows this command can be used
# Removes any row containing a missing value.
# Good when only a few records are affected.
df.dropna()
```

```python
# Fill the mean
# Best for numerical columns
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["CGPA"] = df["CGPA"].fillna(df["CGPA"].mean())
```

```python
# Fill the median
# Best when outliers are present
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].median())
```

```python
# Fill the mode
# Best when categorical data present
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])
df["City"] = df["City"].fillna(df["City"].mode()[0])
```

```python
# Verify the cleaning process
df.isnull().sum()
```

```python
#Check duplicate rows
df.duplicated()
```

```python
# Count duplicate rows
df.duplicated().sum()
```

```python
# Drop the duplicate rows
df = df.drop_duplicates()
```

```python
# Cleaning the text data
# Suppose gender contains Male, male, MALE
# Standardize them using below

df["Gender"] = df["Gender"].str.title()
```

```python
# Remove extra spaces:
df["Name"] = df["Name"].str.strip()
```

```python
# Convert to uppercase
df["Department"] = df["Department"].str.upper()
```

What is an Outlier?

An outlier is an observation that is significantly different from the rest of the data.

```python
# Method 1 to detect outlier
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(y=df["CGPA"])

plt.show()
```

```python
# Method 2 detect outlier usin IQR

Q1 = df["CGPA"].quantile(0.25)
Q3 = df["CGPA"].quantile(0.75)

IQR = Q3 - Q1
```

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
outliers = df[(df["CGPA"] < lower) | (df["CGPA"] > upper)]

outliers
```

```python
df_clean = df[(df["CGPA"] >= lower) & (df["CGPA"] <= upper)]
```
