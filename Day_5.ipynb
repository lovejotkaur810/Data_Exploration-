Feature Engineering

Feature engineering is all about transforming raw data into useful features that help make better predictions with machine learning models. It’s a crucial step that can improve your models and make your data easier to understand.

```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/student_dataset.csv")
```

There are seven common techniques of Feature Engineering

Create

Tranform

Encode

Scale

Bin

Extract

Combine

Technique 1

Feature Creation

Definition

Creating completely new columns from existing columns.

Example- Performance Score= CGPA × Attendance

```python
df["Performance_Score"] = df["CGPA"] * df["Attendance"]
df.head()
```

```python
df["Backlog_Ratio"] = df["Backlogs"]/df["Semester"]
df.head()
```

```python
df["Study_Efficiency"] = df["CGPA"]/df["Study_Hours"]
df.head()
```

Technique 2
Feature Transformation

Explain

Sometimes data exists but in the wrong form.

Tranformation ways:

Square Root

Cube Root

Reciprocal

Normalization

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8,5))

plt.hist(df["Family_Income"].dropna(),
         edgecolor="black")

plt.title("Distribution of Student Family Income")
plt.xlabel("Family_Income")
plt.ylabel("Number of Students")

plt.show()
```

```python
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8,5))

plt.hist(np.log(df["Family_Income"]),
         edgecolor="black")

plt.title("Distribution of Student Family Income")
plt.xlabel("Family_Income")
plt.ylabel("Number of Students")

plt.show()
```

Technique 3: Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df[["CGPA","Attendance"]] = scaler.fit_transform(
df[["CGPA","Attendance"]]
)
df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df["CGPA"] = scaler.fit_transform(df[["CGPA"]])
df.head()
```

Technique 4 Encoding

Computers understand Numbers Not words.

```python
pd.get_dummies(df,
columns=["Department"])
```

```python
# Placed =1, Not Placed = 0
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["Placement_Status"] = encoder.fit_transform(df["Placement_Status"])

print(df[["Placement_Status"]].head())
```

```python
df["Attendance_Level"] = pd.cut(
df["Attendance"],
bins=[0,70,85,100],
labels=["Low",
"Medium",
"High"]
)
df["Attendance_Level"]
```
