```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/student_dataset.csv")
```

```python
# pip install ydata-profiling

from ydata_profiling import ProfileReport
import pandas as pd

profile = ProfileReport(
    df,
    title="Student Dataset Profiling Report",
    explorative=True
)

profile.to_file("student_report.html")
```

```python
# pip install sweetviz

import sweetviz as sv

report = sv.analyze(df)

report.show_html("sweetviz_student_report.html")
```

```python
from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(
    df, test_size=0.2, random_state=42
)
report = sv.compare([train_df, "Train"],
                    [test_df, "Test"])
report.show_notebook()
```
